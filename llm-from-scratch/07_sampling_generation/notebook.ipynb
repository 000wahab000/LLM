## ⚡ Predict Before You Run

**Rule:** Before you execute any code cell, add a markdown cell directly above it and complete this sentence:

> *"I think this will output ___ because ___."*

Run it. If you were wrong — **that gap is the session's real material.** Not the video. Not the notes. The gap.
